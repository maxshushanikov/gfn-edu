## 2. Bedeutung von Python in Unternehmen

Beispiele:

* Google → Automatisierung

## 3. Warum ist Python beliebt?

Hauptgründe:

* einfache Syntax
* riesige Bibliothekslandschaft
* starke Community
* vielseitige Einsatzmöglichkeiten

---

## 4. Unterschied zwischen `.py` und `.ipynb`

`.py`

* reiner Code
* lineare Ausführung
* Softwareentwicklung

`.ipynb`

* Zellen
* interaktive Ausführung
* Visualisierung und Analyse

---

## 5. Unterschied Editor vs IDE

Editor:

* schreibt nur Text

IDE:

* Code schreiben
* debuggen
* ausführen
* Versionierung

---

## 6. Interpretierte vs kompilierte Sprachen

Kompiliert:

* Übersetzung vor der Ausführung

Interpretiert:

* Übersetzung während der Ausführung

---

## 7. CPython vs Cython

CPython:

* Standard-Python-Interpreter

Cython:

* Python mit C-Optimierungen

---

## 8. Wie untersucht man Python Code?

Werkzeuge:

* Linter (z. B. Pylint)
* Debugger
* Logging
* Code-Kommentare
* Dokumentation
