## 2. Theoretische Grundlagen

### 2.1 Geschichte von Python

Python wurde entwickelt von:

**Guido van Rossum**

Er begann die Entwicklung Ende der 1980er Jahre am **Centrum Wiskunde & Informatica (CWI)** in den Niederlanden.

Die erste Veröffentlichung erfolgte:

**1991**

Die Motivation war, eine Sprache zu schaffen, die:

* einfach zu lesen ist.
* leicht zu lernen ist.
* leistungsfähig genug für komplexe Software ist.

---

**Herkunft des Namens**

Viele glauben, der Name kommt von der Schlange.

Das stimmt jedoch nicht.

Der Name stammt von der britischen Comedy-Gruppe:

**Monty Python**

Guido van Rossum war ein großer Fan der Serie **Monty Python’s Flying Circus**.

---

**Python Versionen**

Es gab zwei große Versionen:

**Python 2**

* lange Zeit weit verbreitet
* weniger moderne Architektur

**Python 3**

* modernisierte Version
* bessere Unicode-Unterstützung
* viele Verbesserungen

Seit **2020** ist Python 2 offiziell eingestellt.

Heute wird ausschließlich **Python 3** verwendet.

---

### 2.2 Interpretierte vs. Kompilierte Sprachen

Programmiersprachen unterscheiden sich darin, **wie sie ausgeführt werden**.

---

**Kompilierte Sprachen**

Beispiele:

* C
* C++
* Rust
* Go

Ablauf:

1️⃣ Quellcode wird geschrieben  
2️⃣ Compiler übersetzt gesamten Code in Maschinensprache  
3️⃣ Programm wird ausgeführt  

Beispiel:

```
Code → Compiler → Maschinenprogramm → Ausführung
```

Vorteile:

* sehr hohe Geschwindigkeit
* direkte Maschinenausführung

Nachteile:

* längere Entwicklungszyklen
* erneutes Kompilieren nötig

---

**Interpretierte Sprachen**

Beispiele:

* Python
* JavaScript
* Ruby

Hier wird Code **zur Laufzeit interpretiert**.

Ablauf:

```
Code → Interpreter → Ausführung
```

Der Interpreter liest den Code **Schritt für Schritt**.

---

**Hybridmodell von Python**

Technisch arbeitet Python etwas anders:

Python-Code wird zuerst in **Bytecode** übersetzt.

Dann wird dieser Bytecode von einer **virtuellen Maschine (Python VM)** ausgeführt.

Ablauf:

```bash
Python Code
     ↓
Bytecode (.pyc)
     ↓
Python Virtual Machine
     ↓
Maschinencode
```

---

### 2.3 CPython und Cython

---

**CPython**

CPython ist die **Standardimplementierung von Python**.

Eigenschaften:

* geschrieben in C
* Referenzimplementierung
* offizielle Version von python.org

Wenn Sie Python installieren, installieren Sie **CPython**.

---

**Cython**

Cython ist eine Erweiterung von Python.

Ziel:

Python-Code **beschleunigen**.

Das funktioniert so:

1️⃣ Python-Code wird mit C-Typen erweitert  
2️⃣ Cython übersetzt den Code in **C-Code**  
3️⃣ dieser wird kompiliert  

Beispiel:

```python
cdef int x = 10
```

Cython wird oft in:

* wissenschaftlicher Software
* numerischen Bibliotheken
* Machine Learning

verwendet.
